```python
import zipfile
import pandas as pd

from io import BytesIO

with zipfile.ZipFile("simulation_config.zip", "r") as zf:
    with zf.open("stays.parquet") as f:
        stays_file = BytesIO(f.read())
        occupancy_file = BytesIO(zf.read("occupancy.parquet"))

stays_df = pd.read_parquet(stays_file)
occupancy_df = pd.read_parquet(occupancy_file)
```
